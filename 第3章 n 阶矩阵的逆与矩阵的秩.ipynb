#### 1. 求下列矩阵的逆阵
**思路**：使用伴随矩阵公式 $A^{-1} = \frac{1}{|A|} A^*$（其中 $A^*$ 为伴随矩阵）或初等行变换法 $(A|E) \to (E|A^{-1})$。对于二阶矩阵，可直接用公式 $\begin{bmatrix}a&b\\c&d\end{bmatrix}^{-1} = \frac{1}{ad-bc}\begin{bmatrix}d&-b\\-c&a\end{bmatrix}$。

*   **(1)** $A = \begin{bmatrix} 8 & -4 \\ -5 & 3 \end{bmatrix}$。
    $|A| = 8 \times 3 - (-4) \times (-5) = 24 - 20 = 4 \neq 0$。
    $A^{-1} = \frac{1}{4} \begin{bmatrix} 3 & 4 \\ 5 & 8 \end{bmatrix} = \begin{bmatrix} \frac{3}{4} & 1 \\ \frac{5}{4} & 2 \end{bmatrix}$。
*   **(2)** $A = \begin{bmatrix} \cos\theta & \sin\theta \\ -\sin\theta & \cos\theta \end{bmatrix}$。
    $|A| = \cos^2\theta + \sin^2\theta = 1 \neq 0$。
    $A^{-1} = \begin{bmatrix} \cos\theta & -\sin\theta \\ \sin\theta & \cos\theta \end{bmatrix}$。
*   **(3)** $A = \begin{bmatrix} 2 & 3 & -1 \\ 1 & 2 & 0 \\ -1 & 2 & -2 \end{bmatrix}$。
    $|A| = 2(2\times(-2) - 0) - 3(1\times(-2) - 0) - 1(1\times 2 - 2\times(-1)) = -8 + 6 - 4 = -6 \neq 0$。
    计算伴随矩阵 $A^* = \begin{bmatrix} -4 & 4 & 2 \\ 2 & -5 & -1 \\ 4 & -7 & 1 \end{bmatrix}$。
    $A^{-1} = \frac{1}{-6} \begin{bmatrix} -4 & 4 & 2 \\ 2 & -5 & -1 \\ 4 & -7 & 1 \end{bmatrix} = \begin{bmatrix} \frac{2}{3} & -\frac{2}{3} & -\frac{1}{3} \\ -\frac{1}{3} & \frac{5}{6} & \frac{1}{6} \\ -\frac{2}{3} & \frac{7}{6} & -\frac{1}{6} \end{bmatrix}$。
*   **(4)** $A = \text{diag}(a_1, a_2, \dots, a_n)$，$a_i \neq 0$。
    对角矩阵的逆为对角元素取倒数。
    $A^{-1} = \text{diag}(\frac{1}{a_1}, \frac{1}{a_2}, \dots, \frac{1}{a_n})$。

#### 4. 矩阵可逆性证明
**思路**：构造因式分解，得到 $A \cdot B = E$ 或 $(A+2E) \cdot C = E$ 的形式。
(1) 由 $A^2 - A - 2E = 0$ 得 $A^2 - A = 2E \Rightarrow A(A - E) = 2E \Rightarrow A \cdot \frac{1}{2}(A - E) = E$。故 $A$ 可逆，且 $A^{-1} = \frac{1}{2}(A - E)$。
(2) 分解因式：$(A+2E)(A - 3E) = A^2 - A - 6E$。
因为 $A^2 - A - 2E = 0$，所以 $A^2 - A = 2E$，代入得 $(A+2E)(A - 3E) = 2E - 6E = -4E$。
整理得 $(A+2E)[-\frac{1}{4}(A - 3E)] = E$。故 $A+2E$ 可逆，且 $(A+2E)^{-1} = -\frac{1}{4}(A - 3E)$。

#### 5. 行列式计算
**思路**：利用伴随矩阵性质 $A^* = |A| A^{-1}$ 及行列式性质 $|kA| = k^n|A|$。
$A$ 为三阶矩阵，$|A| = \frac{1}{2}$。
$$|(2A)^{-1} - 5A^*| = \left| \frac{1}{2} A^{-1} - 5 |A| A^{-1} \right| = \left| \frac{1}{2} A^{-1} - \frac{5}{2} A^{-1} \right| = \left| -2 A^{-1} \right|$$
$$= (-2)^3 |A^{-1}| = -8 \times \frac{1}{|A|} = -8 \times 2 = -16$$。

#### 6. 伴随矩阵性质证明
**思路**：利用公式 $AA^* = |A|E$ 和矩阵秩的性质。
(1) 若 $|A|=0$，则 $R(A) < n$。当 $R(A) \le n-2$ 时，$A$ 的所有 $n-1$ 阶子式均为0，故 $A^* = O$，$|A^*|=0$；当 $R(A) = n-1$ 时，$R(A^*) = 1 < n$，故 $|A^*|=0$。
(2) 由 $AA^* = |A|E$，两边取行列式得 $|A| |A^*| = ||A|E| = |A|^n$。若 $|A| \neq 0$，则 $|A^*| = |A|^{n-1}$；若 $|A|=0$，由(1)可知 $|A^*|=0$，等式也成立。故 $|A^*| = |A|^{n-1}$。

#### 7. 解矩阵方程
**思路**：提取公因式，化为 $B = \dots$。
由 $A^{-1}BA = 6A + BA$，右乘 $A^{-1}$ 得 $A^{-1}B = 6E + B$。
移项得 $A^{-1}B - B = 6E$，即 $(A^{-1} - E)B = 6E$。
由于 $A = \text{diag}(\frac{1}{3}, \frac{1}{4}, \frac{1}{7})$，则 $A^{-1} = \text{diag}(3, 4, 7)$。

因此 $B = 6(A^{-1} - E)^{-1} = 6 \text{diag}(\frac{1}{2}, \frac{1}{3}, \frac{1}{6}) = \text{diag}(3, 2, 1)$。

$A = \begin{bmatrix}1&2&2&3\\2&4&t&6\\3&6&6&9\end{bmatrix} \xrightarrow{R_2-2R_1, R_3-3R_1} \begin{bmatrix}1&2&2&3\\0&0&t-4&0\\0&0&0&0\end{bmatrix}$。
显然第一行和第三行非零，若秩为2，则第二行必须非零，即 $t-4 \neq 0 \Rightarrow t \neq 4$。

#### 20. 参数 $k$ 与秩的关系
**思路**：计算行列式 $|A|$。若 $|A| \neq 0$，秩为3；若 $|A|=0$，代入 $k$ 值讨论秩。
$|A| = \begin{vmatrix}1&-2&3k\\-1&2k&-3\\k&-2&3\end{vmatrix} = -6(k-1)(k^2-2) = 0$。
解得 $k=1$ 或 $k=\pm\sqrt{2}$。
(1) 若 $R(A)=1$，所有行成比例，代入 $k=1$ 检验，发现第二行是第一行的$-1$倍，第三行也是第一行的$1$倍，故 $R(A)=1$（成立）。
(2) 若 $R(A)=2$，代入 $k=\sqrt{2}$ 或 $k=-\sqrt{2}$，经初等行变换有一行化为0，故 $R(A)=2$。
(3) 若 $R(A)=3$，即 $k \neq 1$ 且 $k \neq \pm\sqrt{2}$。

#### 21. 秩的不等式证明
**思路**：利用向量组秩的性质，或分块矩阵的性质。
设 $A, B$ 的列向量组分别为 $A_1, \dots, A_n$ 和 $B_1, \dots, B_n$。
$A+B$ 的列向量组为 $A_1+B_1, \dots, A_n+B_n$。
$A+B$ 的列空间 $C(A+B)$ 中的任意向量均可由 $A$ 的列向量与 $B$ 的列向量线性表示，即 $C(A+B) \subseteq C(A) + C(B)$。
根据维数定理，$\dim(C(A+B)) \le \dim(C(A) + C(B)) \le \dim(C(A)) + \dim(C(B))$。
即 $R(A+B) \le R(A) + R(B)$。

#### 22. 分块矩阵的逆
**思路**：设 $\begin{bmatrix}O&A\\B&O\end{bmatrix}^{-1} = \begin{bmatrix}X&Y\\Z&W\end{bmatrix}$，利用逆矩阵定义相乘等于单位阵求解。
$\begin{bmatrix}O&A\\B&O\end{bmatrix} \begin{bmatrix}X&Y\\Z&W\end{bmatrix} = \begin{bmatrix}AZ&AW\\BX&BY\end{bmatrix} = \begin{bmatrix}I&O\\O&I\end{bmatrix}$。
解得：$AZ=I \Rightarrow Z=A^{-1}$；$AW=O \Rightarrow W=O$；$BX=O \Rightarrow X=O$；$BY=I \Rightarrow Y=B^{-1}$。
所以 $\begin{bmatrix}O&A\\B&O\end{bmatrix}^{-1} = \begin{bmatrix}O&B^{-1}\\A^{-1}&O\end{bmatrix}$。
